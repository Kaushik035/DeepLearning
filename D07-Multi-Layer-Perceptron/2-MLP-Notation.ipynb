# MLP Notation

- While making a NN, many weights and biases are involved, so to keep track of them we use a notation system.

- Goal of this lesson:

    1. Kisi vhi NN ko dekh k bata pao ki number of trainable parameters kitne hai usme (Total number of weights + Total number of biases).
    2. How to denote those weights and biases using a notation system.



## 1. Number of trainable parameters in a NN



- In the above pic b/w the input layer and first hidden layer we have 4*3 = 12 weights and 3 biases, so total trainable parameters between input layer and first hidden layer are 12 + 3 = 15.

- Similarly between first hidden layer and second hidden layer we have 3*2 = 6 weights and 2 biases, so total trainable parameters between first hidden layer and second hidden layer are 6 + 2 = 8.

- Similarly between second hidden layer and output layer we have 2*1 = 2 weights and 1 bias, so total trainable parameters between second hidden layer and output layer are 2 + 1 = 3.

- So total trainable parameters in the whole NN are 15 + 8 + 3 = 26.

- So, when we will train this NN, the backpropagation algorithm will update these 26 trainable parameters in such a way that the loss function will be minimized.

## 2. Notation system for weights and biases in a NN

- We will see notation of 3 things:

    1. Notation for biases
    2. Notation for weights
    3. Notation for output of each layer

### 1. Notation for biases

- Bias is denotes as:

    ## $b_{ij}$

- Where i is the layer number and j is the node number in that layer.



### 2. Notation for output of each layer

- Output of each layer is denoted as: (similar to bias)

    ## $O_{ij}$

- Where i is the layer number and j is the node number in that layer from which the output is coming out.



### 3. Notation for weights

- Weight is denoted as:

    ## $w_{ij}^{k}$

- Where $i$ is the node in the previous layer number from which the weight is coming out, $j$ is the node number in the next layer to which the weight is going, and $k$ is the layer number to which the weight is going.
