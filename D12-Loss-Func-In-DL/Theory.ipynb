# What is a Loss Function?



- Higher the value of the loss function, the worse the model is performing.

- Lower the value of the loss function, the better the model is performing.

- Example for linear regression: Mean Squared Error (MSE) as loss function.


- We find the parameters (weights) that minimize the loss function, which is equivalent to finding the best fit line for the data. For that we use optimization algorithms like Gradient Descent.

# Why is Loss Function Important?



- Loss function hume measure karke batate hai ki humare model ki galti kitni hai, or usse hume pata chalta hai ki hume kis direction me jana hai apne model ko improve karne ke liye.

# Lost Function V/S Cost Function

- First thing is some people use them interchangeably, but there is a subtle difference between them.



- Loss function is used to measure the error for a single training example, while cost function is used to measure the error for the entire training dataset.

- **Loss function** is also known as the **Error function**, while **Cost function** is also known as the **Objective function**.



### Let's see different types of loss functions in deep learning...

## 1. **Mean Squared Error (MSE)** / **L2 Loss** / **Squared Loss**



- We take a data point, do forward propagation, get the output, and then we calculate the error by taking the difference between the predicted value and the actual value, then we backpropagate the error to update the weights of the model.

- We took square because if it was not squared, then the positive and negative errors would cancel each other out, and we would not be able to measure the total error correctly.



- Because of the quadratic nature of the MSE loss, more the predicted value is far from the actual value, more the error will be

- **So impact of weight update will be more by the data points which are far from the actual value, and less by the data points which are close to the actual value.** 

### Advantages and Disadvantages of MSE Loss



- It is not robust to outliers, because the error is squared, so the outliers will have a much larger impact on the loss function than the other data points so **line will move towards the outliers and it will not fit the data well.**

- ### <span style="color:orange">**In DL, we can apply MSE loss only when the output neuron has a linear activation function, doesn't matter what activation function we use in the hidden layers, but the output neuron should have a linear activation function** </span>



### 2. **Mean Absolute Error (MAE)** / **L1 Loss** / **Absolute Loss**



### Advantages and Disadvantages of MAE Loss



- We can't directly differentiate the MAE loss function because of the absolute value, so we use sub-gradient descent to optimize the MAE loss function. Which increase the computational cost of training the model.

- ### <span style="color:orange">**If ur data doesn't have outliers, then directly use MSE loss, else use MAE loss**</span>


### 3. **Huber Loss**

- Suppose there are too many outlier points (it means they are basically not outliers they are just a few points that some group of people might have) and all other points are close to the actual value in larger number

- Then using the MSE will give more weight to the outliers, pulling the best fit line towards them, and it will not fit the data well.

- Using the MAE will give equal weight to all the data points, so it will make the line mostly stay close to the actual value, but then the outlier groups will be ignored, and the line will not fit the overall data well.

- So, Huber loss deals with this problem by combining the advantages of both MSE and MAE loss functions to get a line in between the two



- So agar the error is less than delta (means it's not an outlier), then we use the MSE loss function, and if the error is greater than delta, then we use the MAE loss function.

- ###  <span style="color:orange">So, it performs well when our data have many outliers</span>

### 4. **Binary Cross-Entropy Loss** / **Log Loss**



- ### <span style="color:orange">It is important to note that we can only use sigmoid activation function in the output layer when we are using binary cross-entropy loss function, though we can use any activation function in the hidden layers, but the output neuron should have a sigmoid activation function.</span>

### Advantages 

- It is differentiable, so we can directly use gradient descent to optimize the binary cross-entropy loss function.

### Disadvantages

- Can have multiple local minima, which can make it difficult to optimize the loss function and find the global minimum.

- Not intuitive that how the loss function is working


### 5. **Categorical Cross-Entropy Loss** / **Softmax Loss**



- **The number of output neurons will be equal to the number of classes in the dataset, and we will use softmax activation function in the output layer to get the probabilities of each class.**

- If we use Categorical Cross-Entropy Loss for a multi-class classification problem, then we have to one-hot encode the target labels

- If we use Sparse Categorical Cross-Entropy Loss for a multi-class classification problem, then we don't have to one-hot encode the target labels, we can directly use the class labels as target labels if thery were numberic class labels, if they were string class labels, then we have to encode them to numeric class labels before using sparse categorical cross-entropy loss function. (Eg. Yes:1 No:2 Maybe:3)



- So here in the loss function formula, if for a row the actual class is 1, then the loss function will only consider the predicted probability of class 1, and if for a row the actual class is 2, then the loss function will only consider the predicted probability of class 2, and if for a row the actual class is 3, then the loss function will only consider the predicted probability of class 3. And inside the log function, we take the predicted probability of the actual class

- SO **if you have lots of classes in your dataset**, then use Sparse Categorical Cross-Entropy Loss, because it is computationally more efficient than Categorical Cross-Entropy Loss, because in Categorical Cross-Entropy Loss we have to one-hot encode the target labels, which increases the computational cost of training the model, while in Sparse Categorical Cross-Entropy Loss we don't have to one-hot encode the target labels, so it is computationally more efficient than Categorical Cross-Entropy Loss.
