# Types of Neural Networks

## 1. Multilayer Perceptron (MLP) / Artificial Neural Network (ANN)

- Specially useful in supervised learning tasks (classification and regression)
- Consists of a series of layers: input layer, hidden layers, and output layer
- Can capture non-linear relationships in data

## 2. Convolutional Neural Networks (CNNs)



- Primarily used for image and video recognition tasks

## 3. Recurrent Neural Networks (RNNs)

- ANN and CNN are feedforward networks, meaning that data flows in one direction from input to output. RNNs, on the other hand, have connections that form directed cycles, allowing them to maintain a 'memory' of previous inputs.

- This makes RNNs particularly effective for tasks involving sequential data, such as time series analysis, natural language processing, and speech recognition. 



- A very famous RNN architecture is the Long Short-Term Memory (LSTM) network, which is designed to overcome the vanishing gradient problem that can occur in traditional RNNs when dealing with long sequences.

## 4. Autoencoders

- Used when we want to compress the input data, without losing important information. They consist of an encoder that compresses the input into a latent space representation, and a decoder that reconstructs the original input from this representation.



# 5. Generative Adversarial Networks (GANs)

- Consist of two neural networks, a generator and a discriminator, that are trained together in a game-theoretic framework. The generator creates fake data samples, while the discriminator tries to distinguish between real and fake samples. 

- When generator fools the discriminator, we will tell that the particular output(let say image) is a auto generated image (i.e. imaginary image). GANs are widely used in image generation, music generation, and other creative applications.

# Why Deep Learning is gaining popularity?

1. Better hardware availability: The availability of powerful GPUs and TPUs has made it possible to train deep neural networks efficiently.

2. Labeled data availability: The increase in labeled data, especially in fields like computer vision and natural language processing, has enabled the training of deep learning models.

3. Libraries like TensorFlow, PyTorch, and Keras have made it easier for researchers and practitioners to build and deploy deep learning models.
