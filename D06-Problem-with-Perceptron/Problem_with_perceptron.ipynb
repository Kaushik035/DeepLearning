# Problems with Perceptron

- Perceptron didnt't became much famous in DL because it can only work well with linear data. It can't solve non-linear problems.

- Doesn't matter how many times you train a perceptron, it will never be able to solve non-linear problems. As ultimately its a linear model.

- Let's see this with the help of some datasets. We will be working with 3 datasets: OR, AND and XOR.

- OR and AND datasets are linear datasets. XOR dataset is a non-linear dataset as it cannot be separated into two classes by a single straight line. We will see how perceptron performs on these datasets.



```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# We will be working with 3 datasets: OR, AND and XOR.
#  These are the most basic datasets in ML. They are used to understand the concept of linear and non-linear data.

or_data = pd.DataFrame()
and_data = pd.DataFrame()
xor_data = pd.DataFrame()
```

```python
or_data['input1']=[1,1,0,0]
or_data['input2']=[1,0,1,0]
or_data['output']=[1,1,1,0]
```

```python
and_data['input1']=[1,1,0,0]
and_data['input2']=[1,0,1,0]
and_data['output']=[1,0,0,0]
```

```python
xor_data['input1']=[1,1,0,0]
xor_data['input2']=[1,0,1,0]
xor_data['output']=[0,1,1,0]
```

```python
and_data
```

```python
sns.scatterplot(
    x=and_data['input1'],
    y=and_data['input2'],
    hue=and_data['output'],
    s=200
)
```

```python
or_data
```

```python
sns.scatterplot(
    x=or_data['input1'],
    y=or_data['input2'],
    hue=or_data['output'],
    s=200
)
```

```python
xor_data
```

```python
sns.scatterplot(
    x=xor_data['input1'],
    y=xor_data['input2'],
    hue=xor_data['output'],
    s=200
)
```

```python
from sklearn.linear_model import Perceptron
clf1=Perceptron()
clf2=Perceptron()
clf3=Perceptron()
```

```python
clf1.fit(and_data.iloc[:,0:2].values,and_data.iloc[:,-1].values)
clf2.fit(or_data.iloc[:,0:2].values,or_data.iloc[:,-1].values)
clf3.fit(xor_data.iloc[:,0:2].values,xor_data.iloc[:,-1].values)
```

```python
clf1.coef_
```

```python
clf1.intercept_
```

```python
x=np.linspace(-1,1,5)
y=-x+1
```

```python
plt.plot(x,y)
sns.scatterplot(x=and_data['input1'], y=and_data['input2'], hue=and_data['output'], s=200)
```

```python
clf2.coef_
```

```python
clf2.intercept_
```

```python
x1=np.linspace(-1,1,5)
y1=-x+0.5
```

```python
plt.plot(x1,y1)
sns.scatterplot(x=xor_data['input1'], y=xor_data['input2'], hue=xor_data['output'], s=200)
```

- So, you can see for OR and AND dataset, we can easily draw a line to separate the data points. But for XOR dataset, we can't draw a line to separate the data points. This is the problem with perceptron. It can only work well with linear data. It can't solve non-linear problems.

```python
clf3.coef_
```

```python
clf3.intercept_
```
