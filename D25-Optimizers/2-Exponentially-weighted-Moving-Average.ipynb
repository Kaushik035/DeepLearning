# Exponenetially weighted moving average (EWMA)

- The exponentially weighted moving average (EWMA) is a technique used to smooth time series data by giving more weight to recent observations and less weight to older observations. It is commonly used in optimization algorithms to track the gradient of the loss function over time.



- The black curve is the EWMA and the red line is the average of all the points in the above graph. The EWMA is more responsive to recent changes in the data, while the simple average is more stable and less sensitive to recent changes.

- EWMA is used to show the trend of the data and to reduce the noise in the data. 

## We have to remember 2 things about EWMA working:

1. The recent points are more important than the older points. So we give more weight to the recent points and less weight to the older points.
2. The weight of each point decreases as the point gets older.



- Let's see how to calculate the EWMA of a the temperature data in the above pic



- $V_t = \beta V_{t-1} + (1 - \beta) x_t$
- $V_t$ is the EWMA at time step t, $V_{t-1}$ is the EWMA at time step t-1, $x_t$ is the value of the data point at time step t, and $\beta$ is a smoothing factor that determines how much weight to give to the previous EWMA value.

- For $V_0$, we typically initialize it to the first data point, i.e., $V_0 = x_0$. But we can also initialize it to 0, i.e., $V_0 = 0$. However, initializing it to $x_0$ is more common in practice.



# Interpretation of $\beta$

- In EWMA, $\beta$ is the **decay (momentum) factor**. The update is $v_t = \beta \, v_{t-1} + (1 - \beta)\, \theta_t$.

- For $\beta = 0.9$:

$$
\frac{1}{1 - \beta} = \frac{1}{1 - 0.9} = \frac{1}{0.1} = 10
$$

- So, the EWMA at time step $t$ can be thought of as **approximately a normal average of the last 10 data points**.




- The lower the value of $\beta$, the **more weight goes to recent points** and the less to older points. The curve follows the data closely but is **less smooth**.

- The higher the value of $\beta$, the **more weight goes to older points** and the less to recent points. The curve is **smoother** but **reacts slowly** to changes. (As you can see in the above graph for $\beta = 0.98$ its value is always lower because it is giving more weight to older points and less to recent points.)

- Sweet spot for $\beta$ is usually 0.9.

# Mathematical Intuition  of older the points gets less important it becomes



- So as you can see int the above pic that, the weight of the older points decreases exponentially as the point gets older. The weight of the point at time step $t-k$ is $\beta^k$. So, the weight of the point at time step $t-1$ is $\beta^1$, the weight of the point at time step $t-2$ is $\beta^2$, and so on.

- And since $\beta$ is less than 1, the weight of the older points decreases exponentially as the point gets older. This is why it is called **exponentially weighted moving average**.
