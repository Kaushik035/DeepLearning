# Role of Optimizers in Neural Networks

-  In a neural network, the optimizer is responsible for updating the weights of the network during training. The goal of the optimizer is to minimize the loss function by adjusting the weights in a way that reduces the error between the predicted output and the actual output.

- The optimizer we use in Deeplearning is Gradient Descent.



### Types of Optimizers we studied till now



- But there are some challanges with using Gradient Descent as an optimizer. So we will study some advanced optimizers which are used in practice to improve the training speed of neural networks.

### Challenges with Gradient Descent

1. **Choosing the right learning rate**: If the learning rate is too small, the training will be slow. If the learning rate is too large, the training may diverge and not converge to a minimum.
    - The solution to this problem is to use learning rate schedules or adaptive learning rate methods.
    - But **problem with learning rate schedules** is that we need to predefine the learning rate schedule (LRS) before training. But a LRS that works well for one dataset may not work well for another dataset. So we need to tune the learning rate schedule for each dataset. This is a time consuming process.

2. For each learning parameter, we **can't use different learning rates** in Gradient Descent. But in practice, we may want to use different learning rates for different learning parameters.

3. The biggest problem with Gradient Descent is that it can get **stuck in local minima**. This can happen when the loss function is non-convex, which is often the case in deep learning. In such cases, the optimizer may converge to a suboptimal solution.

4. Getting **stuck in saddle points**. A saddle point is a point in the loss function where the gradient is zero, but it is not a minimum or maximum. In such cases, the optimizer may oscillate around the saddle point and not converge to a minimum.
