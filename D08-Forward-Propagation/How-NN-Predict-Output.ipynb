The outputs of each layer are also called as the activations of that layer, because these are the values which will be passed to the next layer as input.

- Let the Input be Activation of layer 0 ($a^{(0)}$), then the output of layer 1 will be the activation of layer 1 ($a^{(1)}$) and the output of layer 2 will be the activation of layer 2 ($a^{(2)}$) and so on.

- So ultimately we are doing this:

1. $a^{(1)} = \sigma(W^{(1)}a^{(0)} + b^{(1)})$

2. $a^{(2)} = \sigma(W^{(2)}a^{(1)} + b^{(2)})$

3. $a^{(3)} = \sigma(W^{(3)}a^{(2)} + b^{(3)})$

- And the final output of the NN will be $a^{(3)}$ which is the activation of the output layer.

- All together we can write the output of the NN as:

$$\hat{y} = a^{(3)} = \sigma(W^{(3)}\sigma(W^{(2)}\sigma(W^{(1)}a^{(0)} + b^{(1)}) + b^{(2)}) + b^{(3)})$$
