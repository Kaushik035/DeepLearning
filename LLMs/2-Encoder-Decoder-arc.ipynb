# Detour 



- Till now, you see back the journey:

1. When we had **tabular data** and we wanted to predict something out of it, We used ANNs (Artificial Neural Networks)

2. For **image data**, it can be showed using grid and ANN were not sufficient to capture the spatial features of images. Hence, we moved to Convolutional Neural Networks (CNNs) which are designed to process grid-like data such as images.

3. When we encountered **sequential data**, such as time series or natural language, we realized both ANN and CNNs were not able to capture the temporal dependencies in the data. This led us to Recurrent Neural Networks (RNNs), which are specifically designed to handle sequential data by maintaining a hidden state that captures information from previous time steps.
    - **LSTM (Long Short-Term Memory)** networks were introduced to address the vanishing gradient problem in RNNs, allowing them to capture long-term dependencies more effectively.
    - **GRU (Gated Recurrent Unit)** networks were also developed as a simpler alternative to LSTMs, providing similar performance with fewer parameters.

4. Now for Sequence to sequence data, we need to find a NN architecture that can solve seq-to-seq problems, where both the input and output are sequences. This is where the Encoder-Decoder architecture comes into play, which is commonly used in tasks like machine translation, text summarization, and speech recognition.

# Why sequence to sequence problems are challenging?

- **Variable Lengths**: The input and output sequences can have different lengths, making it difficult to align them directly.

- Through LSTMs we somewhat learned to handle variable length inputs, but the challenge arises when we need to generate outputs of variable lengths as well.
