# Regularization in Deep Learning

- Regularization is a technique used in deep learning to prevent overfitting, which occurs when a model learns the training data too well and fails to generalize to new, unseen data.

# Why does NN overfit?

- Because of its complex structure it tries to capture every pattern in the training data, including noise and outliers, leading to poor performance on test data.











- You can see with increasing complexity of the model above, NN starts to overfit the training data, leading to a decrease in performance on the validation set.



- In deep learning mostly L2 regularization is used, it gives good results in most cases.



# L2 Regularization

- In regularization we just add a penalty term to the loss function, which discourages the model from learning overly complex patterns.



- In L2 regularization, we add the sum of the squares of the weights to the loss function as a **penalty term** 

- $\lambda$ = regularization hyperparameter that controls the strength of the penalty. A higher value of $\lambda$ means more regularization, which can help prevent overfitting but may also lead to underfitting if set too high.

- $n$ = number of rows in the training data. (The 2 multiplication by 2 is for mathematical convenience when taking the derivative.)

- The work of the penalty term is to shrink the weights towards zero, which reduces the model's complexity and helps prevent overfitting.

# Intuition behind Regularization

- Why on add adding penalty term to the loss function, the weights shrunks towards zero?



See the bottom right corner in the image above, then new weight update formula, because of multiplication of $W_{old}$ with $(1- \eta \lambda)$, the $W_{old}$ will decrease in value, so $W_{new}$ will be smaller when regularization is applied than when it is not applied. This is how the weights shrink towards zero.

- With each epoch, the weights are decreasing in value, and the model is less likely to overfit the training data.

- This $(1- \eta \lambda)$ term is called the **weight decay**. It is always less than 1, which means that the weights will always decrease in value with each update.



- In L2 regularization Weights are shrunk towards zero, but they never become exactly zero. This means that all features are still used in the model, but their influence is reduced.

- But in L1 regularization, some weights can become exactly zero, which means that some features are completely ignored by the model. This can be useful for feature selection, as it allows the model to focus on the most important features.
