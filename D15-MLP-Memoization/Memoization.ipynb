# Memoization

- Memoization is a technique used to optimize the performance of recursive algorithms by storing the results of expensive function calls and returning the cached result when the same inputs occur again. This can significantly reduce the time complexity of algorithms that have overlapping subproblems, such as Fibonacci sequence calculation.

# Memoization used in MLP

- The above example is for 2 hidden layers, but in real life we can have more than 2 hidden layers. So, the dependence will be more complex and we will have to consider all the paths through which the output of the first hidden layer is passing and then apply chain rule to find the derivative of loss with respect to the parameters of the first hidden layer.

- So, you can see a pattern that the derivatives that we calculated in the 2nd hidden layer are being used in the calculation of derivatives in the 1st hidden layer. So, we can store the derivatives calculated in the 2nd hidden layer and use them in the calculation of derivatives in the 1st hidden layer. This is called memoization.

- Similarly we can extend this to more hidden layers and store the derivatives calculated in each hidden layer and use them in the calculation of derivatives in the previous hidden layer. This way we can optimize the performance of our algorithm and reduce the time complexity.

- ### So, it makes training of the NN faster by avoiding redundant calculations of derivatives.
