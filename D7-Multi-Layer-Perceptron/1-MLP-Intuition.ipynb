# The Problem with Perceptrons

- It can't capture non-linearity in data.



- So, hume ek esa algorithm banana hai jo kisi vhi type k non linearity capture kar paye, jo perceptron capture nahi kar pata.

- But the challenging part is hume Perceptrons use karke hi wo algorithm banana hai

# Perceptron with Sigmoid Activation Function

- So, in this lecture we will use sigmoid function as the activation function instead of step function and will take binary cross entropy as the loss function to explain te Multi Layer Perceptron (MLP) architecture.



- So, suppose after training we got w1 = 5,w2 =10 and b = 3
- So we are predicting wether a student with cgpa = 8.7 and iq = 8.7 will get placed or not.

- So, we will calculate the value of z first

$z = w1*x1 + w2*x2 + b$

- Now, we will apply sigmoid function on z to get the predicted probability of the student getting placed.

$y_{pred} = \sigma(z) = \frac{1}{1 + e^{-z}} = 0.8 (let's say)$

- So probability of getting placed is 0.8 and probability of not getting placed is 0.2

- So, parallel to to best fit line, a gradient of Probability of getting placed appears because of sigmoid function.



- So for every student we will get a probability of getting placed and not getting placed.

# Abstract idea of MLP

- We took 2 perceptron and trained on the same data and got different weights and bias for both the perceptrons.



- What if we can superimpose the 2 decision boundaries of the 2 perceptrons and get a non linear decision boundary.



- If we apply smoothening function on the superimposed decision boundary then we will get a non linear decision boundary.



# Mathematical implementation of the above idea to implement MLP



- Take a student (a data point) , calculate probability of getting placed for both the perceptrons (0.7 and 0.8 here)

- Now add them and get 1.5, but we want the output to be between 0 and 1, so we will apply sigmoid function on 1.5 to get the final output.

$y_{final} = \sigma(1.5) = 0.82$

- So, the final output is 0.82 which is the probability of getting placed for that student after superimposing the decision boundaries of both the perceptrons.

- Now, do this for all the students and we will get a non linear decision boundary.

- So we are using linear combination of the outputs of the perceptrons and then applying sigmoid function on it to get the final output.

## Now suppose we want a different decision boundary, where we want to give more weightage to the output of the first perceptron than the second perceptron.



- So, do weighted addion of the outputs of the perceptrons and then apply sigmoid function on it to get the final output.



- Here weight of the first perceptron is 10 and weight of the second perceptron is 5 

- Now, do this for all the students and we will get a different non linear decision boundary where impact of the first perceptron is more than the second perceptron.

## We can also add bias to the weighted addition of the outputs of the perceptrons



- So, in a way we can cay that the superimposed perceptron (the red box drawn in the below image) is also a perceptron with sigmoid activation function, inputs as the outputs of the 2 perceptrons and weights as the weights of the weighted addition and bias as the bias of the system (superimposed perceptron).









- So the above is a Multi layer perceptron

- So we take linear combination of the outputs of the perceptrons and then apply sigmoid function on it to get the final output to capture non linearity in the data.

# Let's see if we make the above architecture more complex by adding more perceptrons in the hidden layer and more hidden layers then what will happen.

- Let's see how we can change our neural network architecture to have more flexibility in capturing non linearity in the data.

- Neural Network Architecture means, how the nodes (perceptrons) are connected to each other, how many hidden layers we have, how many perceptrons we have in each hidden layer, what activation function we are using in each layer, what is the loss function we are using to train the model etc.

- We can change the architucture of NN using the following 4 ways:

## 1. Adding nodes in the hidden layer





- So the more the nodes you add in the hidden layer, more will be helpful in creating more complex non linear decision boundary

## 2. Adding nodes in Input layer

- This happen when we add more features to the data, so we will have more input nodes in the input layer.

- For 3 features we will have 3 input nodes in the input layer and the desicion boundary will be in 3D space (it willbe a plane in 3D space) and for 4 features we will have 4 input nodes in the input layer and the desicion boundary will be in 4D space (it will be a hyperplane in 4D space) and so on.





## 3. Adding nodes in the output layer

- Generally we do this when we have multi class classification problem, where we have more than 2 classes to predict.



- Output mei jis vhi class ka probability sabse jada hoga usko predict karenge

# Add number of hidden layers (Deep Neural Network)

- Jb humare paas bohot complex data hota hai jisme bohot zyada non linearity hoti hai to hum ek hidden layer se zyada hidden layers add kar dete hai taki hum us data ke non linearity ko better capture kar sake.



- Also a point to note that the perceptron of initial hidden layer will have a linear decision boundary and as we add more hidden layers the decision boundary of perceptrons in those layers will become more and more non linear and complex as we go deeper in the network.

- **That's why NN is knowns as Universal Function Approximator, because it can approximate any function (linear or non linear) given enough hidden layers and hidden units in each layer.**
