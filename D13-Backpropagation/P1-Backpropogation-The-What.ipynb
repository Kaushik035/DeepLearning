# Backpropogation

- It is an algorithm which is used to train neural networks.

- Training means finding the optimum value of weights and biases of a neural network to minimize the loss function

# Steps in Backpropagation:



- So,based on the initalized weights and biases, we will do a forward pass to calculate the output of the neural network for a given input:



- Suppose for 1st point $y_{pred}$ came 18 lpa, but actual value is 3 lpa, so this error is due to random initialization of weights and biases, so we need to update the weights and biases to minimize this error



- So, $\text{Error} = (\text{Actual value} - \text{Predicted value})^2 = (3 - 18)^2 = (-15)^2 = 225$ 

- Now, in  $ L = \frac{1}{n} \sum_{i=1}^{n} (y_{i} -  \hat{y}_{i})^2 $  we can only change $$\hat{y}$$ to minimize the loss function (humare case mei for the above example $\hat{y}$ ko ghatana padega )

- ### But $\hat{y}$ ko badhana hai ya ghatana hai, ye kis cheez pe depend krega?

$$ \hat{y} = O_{21} = W_{11}^{(2)} O_{11} + W_{21}^{(2)} O_{21} + b_{21}  $$

- So, $\hat{y}$ depends on 5 things: $W_{11}^{(2)}$, $W_{21}^{(2)}$, $b_{21}$, $O_{11}$ and $O_{21}$



- $W_{11}^{(2)}$, $W_{21}^{(2)}$, $b_{21}$ can be changed by us, but $O_{11}$ and $O_{21}$ are the output of the previous layer, so we cannot change them directly, we can only change them indirectly by changing the weights and biases of the previous layer



- So its kind of a complex hirerarchy, we need to go back layer by layer to update the weights and biases,  and thats why this is called Backpropagation





- Now we have to claculate derivative of loss function with respect to all the weights and biases ($W_{11}^{(2)}$, $W_{21}^{(2)}$, $b_{21}$, $W_{11}^{(1)}$, $W_{12}^{(1)}$, $W_{21}^{(1)}$,$W_{22}^{(1)}$, $b_{11}$, $b_{21}$) so that we can update the weights and biases in the direction of minimizing the loss function

- So, a total of 9 derivatives we have to calculate

# Calculating the derivatives:

- We need to minimize the loss function $->$ Loss function depends on $\hat{y}$ $->$ $\hat{y}$ depends on Weights and Biases

- ### But then why are we claculating the derivatives of Loss function with respect to weights and biases?

    - $\frac{\partial L}{\partial W_{11}^{(2)}}$ means how much the loss function will change if we change the weight $W_{11}^{(2)}$ by a small amount

    - But $\frac{\partial L}{\partial W_{11}^{(2)}}$ is not directly related to the loss function, it is related to $\hat{y}$, and $\hat{y}$ is related to Loss function

    - So, its a indirect relationship, if we change $W_{11}^{(2)}$ by a small amount, then $\hat{y}$ will change by a small amount, and if $\hat{y}$ changes by a small amount, then the loss function will change by a small amount


- So, we can't directly calculate $\frac{\partial L}{\partial W_{11}^{(2)}}$, until we calculate $\frac{\partial L}{\partial \hat{y}}$ and then multiply it by $\frac{\partial \hat{y}}{\partial W_{11}^{(2)}}$

$$
\frac{\partial L}{\partial W_{11}^{(2)}} = \frac{\partial L}{\partial \hat{y}} \cdot \frac{\partial \hat{y}}{\partial W_{11}^{(2)}}
$$
