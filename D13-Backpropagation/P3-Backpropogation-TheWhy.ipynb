- Revising the algorithm of backpropagation:




- The main step is the below step where we calculate the gradients of the loss w.r.t weights and bias using chain rule:


- We will understand why on subtracting/adding the gradients to the weights and bias we are moving towards the minima of the loss function

# <span style="color:orange">Some Concepts to understand before we understand why backpropagation works:</span>

## 1. Loss function is a function of all trainable parameters (weights and bias)



- Assume NN like a box with 9 knobs (weights and bias) and we are trying to find the right combination of these knobs to get the minimum loss


## 2. Concept of Gradient

- When a function has multiple variables, the gradient is a vector that contains the partial derivatives of the function with respect to each variable.



- Geometrically we are calculating the slope of a function with respect to each parameter (weight and bias) 

## 3. Intuition/Concept  of Derivative

- Rate of change of a function with respect to its variable is called derivative



## 4. Concept of minima

 

# Now let's understand why backpropagation works?

- We will take 2 assuptions to understand this:

    1. learning_rate  = 1 
    2. Out of the 9 parameters (weights and bias) only 1 parameter is contributing to the loss function, rest all are not contributing to the loss function (i.e. loss function is only a function of 1 parameter)

    

- For the loss function be only $f(b_{21})$

- If $\frac{\partial L}{\partial b_{21}} > 0$, then $b_{21}$ should be decreased to move towards minima, and if $\frac{\partial L}{\partial b_{21}} < 0$, then $b_{21}$ should be increased to move towards minima





- So, the negative sign in the update rule of weights and bias is there to move towards minima of the loss function for both the cases when the gradient is positive and negative

- Geometrically:



- So the update rule:
$$b_{21} = b_{21} - \frac{\partial L}{\partial b_{21}}$$

is moving towards minima of the loss function by increasing or decreasing the value of $b_{21}$ by the value of slope at that point (which is the value of gradient at that point)

## Effect of learning rate:

- It ensure smoother steps towards minima, and also helps to avoid overshooting the minima

- For larger learning rates we are taking larger steps towards minima, and for smaller learning rates we are taking smaller steps towards minima

## Till when we run our backpropagation algorithm?

- Till we get convergence.

### What is convergence?



- Convergence is when $W_{new} = W_{old}$, meaning the weights are no longer changing significantly.

- Meaning the slope of the loss function is almost zero, and we are at minima of the loss function

- But in general we set number of epochs and run the backpropagation algorithm for that many epochs, and for mostly epochs = 1000 is a good choice to get convergence in most of the cases
