- There are some heuristics (jugad) to initialize the weights of a neural network, which work well in practice. Those are:
    1. Xavier/ Glorot initialization
        - Generally used when using tanh/sigmoid activation functions
    2. He initialization
        - Generally used when using ReLU activation functions

    


# Xavier/ Glorot initialization (Normal Distribution based)

- Xavier initialization is a weight initialization technique that helps to keep the variance of the activations and gradients across layers in a neural network approximately the same. This is important for preventing the vanishing or exploding gradient problem during training.

- We can initialize the weights of a layer with $n_{in}$ input neurons and $n_{out}$ output neurons using the following formula:

$$ np.random.randn(n_{in}, n_{out}) * \sqrt{\frac{1}{n_{in}}} $$

Where $n_{in}$ is the number of input neurons and $n_{out}$ is the number of output neurons. This will ensure that the variance of the activations and gradients is approximately the same across layers.



- The intuition is that, $z = w \cdot x$ is a linear combination of the inputs. So more the number of inputs, more will be the variance of $z$. Hence we need to scale down the weights by $\sqrt{n_{in}}$ to keep the variance of $z$ same across layers.

- So we divide the weights by $\sqrt{n_{in}}$ to keep the variance of $z$ same across layers. This is called Xavier initialization.

- Only use it for **tanh/sigmoid** activation functions.

## He initialization (Normal Distribution based)

- We can initialize the weights of a layer with $n_{in}$ input neurons and $n_{out}$ output neurons using the following formula:

$$ np.random.randn(n_{in}, n_{out}) * \sqrt{\frac{2}{n_{in}}} $$



- Use it for **ReLU** activation functions.
