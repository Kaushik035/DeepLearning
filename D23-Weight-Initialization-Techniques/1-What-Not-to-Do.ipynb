# Why Weight initialization is important?

- Neural networks in the following way:


- The first step, of initializing parameters, is very important.

- If not done properly may lead to the following problems:
    - Vanishing gradients problem
    - Exploding gradients problem
    - Slow convergence

- It is so important that during 2000's, DL research was stopped because of the vanishing gradients problem.

- Then the researchers found 2 culprits for the vanishing gradients problem:
    1. Activation functions (sigmioid)
    2. Wrong Weight initialization

# What not to do?

## Case 1: Initialize all weights to 0



### Case 1.1 The activation function is ReLU in the hidden layer







- So if you are using ReLU as the activation function in the hidden layer, then initializing all weights to 0, then the gradient will be 0 and the weights will not be updated. Hence the model will not learn anything.

### Case 1.2 The activation function is Tanh in the hidden layer

$$ a_{11} = tanh(z_1) = \frac{e^{z_1} - e^{-z_1}}{e^{z_1} + e^{-z_1}} $$

- Since $z_1 = 0$, $a_{11} = 0$ and hence the gradient will also be 0. So the weights will not be updated. Hence the model will not learn anything.

See [non-zero_initialization_relu](non-zero_initialization_relu.ipynb) for code implementation of this case. But it have case for initialization with a non-zero constant value. So you can edit the code to initialize with 0 and see the results.

### Case 1.3 The activation function is Sigmoid in the hidden layer

$$ a_{11} = \sigma(z_1) = \frac{1}{1 + e^{-z_1}} $$



- Since, $z_1 = 0$, $a_{11} = 0.5$ 





- All the pink and blue boxes will be same value according to the box

- So only the last term will vary in the gradient





- Hence from the above pic, you can see, all the weight coming out of the same nodes will be updated with the same value. So the weigts coming out of the same node will be same. 
- Hence the 2 neurons in the hidden layer will be same.

- SO it cant capture the non-linearities in the data. Hence will just behave like a linear model (perceptron).



You can see the [zero_initialization_sigmoid](zero_initialization_sigmoid.ipynb) for code implementation of this case.

## Case 2: Initialize with a non-zero constant value

- For any activation function used:



- You can see as $z_{11}$ = $z_{12}$, hence all the activations of the neurons in the hidden layer will be same.



- Hence you can see that all the weights coming out of the same node will be updated with the same value. So the weigts coming out of the same node will be same.

- So all the neurons in the hidden layer will be same. Hence it cant capture the non-linearities in the data. Hence will just behave like a linear model (perceptron).

## Case 3: Random Initialization

### Case 3.1: Random Initialization with small values



- Initial setup is we have a data set withh 500 features and 1000 rows and a target variable, there are 3 hidden layers with 500 neurons each and a output layer with 1 neuron.

- The values of all the 500 features follows a normal distribution with mean 0 and standard deviation 1.



- we initialize weights as above


- So, $z$ is a small number

- For **tanh/sigmoid** activation function, the activation value of $z$ will be smaller



- See the distribution of outputs after each layer in the above pic, as the distribution is shown at the bottom of each layer.

- Now, since the activation values are small, the gradient will also be small. Hence the weights will be updated with a small value. Hence the weights will not be updated properly and the model will not learn properly. (Vanishing gradients problem)

- For **ReLU** activation function, here also the activation value of $z$ will be smaller, but since ReLU is a non-saturating activation function there will be less vanishing gradients problem. 

- But model learing will be slow because the weights are updated with small values. Hence it will take more epochs to converge.

### Case 3.2: Random Initialization with large values

- Setup is same as previous case 3.1
- Weights are initialized with large values between [0,1], yes in ML this are considered large values.



- Because, since we take $z = w_1x_1 + w_2x_2 + ... + w_{500}x_{500}$, and $w_i$ is [0,1], $z$ will be a large number.




- So when we use **tanh/sigmoid** activation function, the activation value of $z$ will be saturated as $z$ is a large number

- Because of this saturation, two things are possible:
    1. Training will be slow
    2. There might be problem of vanishing gradients in worst case scenario



- For **ReLU** activation function, larger the value of $z$, larger will be the activation value. Hence the gradient will also be large.

- So we will take larger steps in the gradient descent resulting to unstable gradient descent. And might overshoot the minima. Hence the model will not converge properly. (**Exploding gradients problem**)
