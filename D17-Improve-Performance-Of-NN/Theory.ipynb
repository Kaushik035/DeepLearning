# Fine-tuning the hyperparameters 

## 1. No. of hidden layers

- Instead of taking 1 hidden layer with more neurons, we can take more hidden layers with fewer neurons in each layer. This can help the model to learn more complex patterns in the data and improve its performance.

- Because DL use representation learning, where each layer learns a different representation of the data, having more hidden layers can allow the model to learn more complex and abstract representations, which can lead to better performance.

- The initial layers can learn low-level features (like edges in images), while deeper layers can learn higher-level features (like shapes and objects in images). This hierarchical learning can improve the model's ability to generalize and make accurate predictions.

- Above point se or ek faida hota hai: **Transfer Learning**



- Like Suppose we trained a model for human face recognition, then we can use the initial layers of that model (which have learned to detect low-level features like edges and textures) and fine-tune the later layers for a different but related task, such as monkey face recognition. This can save time and computational resources while still achieving good performance.

- ### <span style="color:orange">But how many layers should we have?  </span>

    - Start with a simple architecture and gradually increase the number of layers while monitoring the model's performance

    - If **Overfitting** occurs, then stop at that point and do not add more layers.

## 2. Number of neurons in each hidden layer

- For input and output layers, the number of neurons is determined by the number of features in the input data and the number of classes in the output, respectively.

- For hidden layers, there is no fixed rule for determining the number of neurons. It often requires experimentation and tuning to find the optimal number of neurons for a given problem.

- Earlier some researchers suggested that the number of neurons in a hidden layer should follow a pyramid structure, where the number of neurons decreases as we go deeper into the network. The logic was that the initial layers should have more neurons to capture low-level features, while deeper layers can have fewer neurons to capture higher-level features that is formed from the low-level features.

- However experiments have shown that this is not a strict rule even if all the layers have the same number of neurons, the model can still learn effectively

- ### Just you should take care the nodes are sufficient to capture the features



- As in the above pic, if you keep only one node in the first hidden layer, it will not be able to capture all the features of the input data and in the next layer you will never be able to recover the features that are not captured in the first layer, so you will end up with a very poor model.

- <span style="color:orange">So, **NODES SHOULD BE MORE THAN REQUIRED TO CAPTURE THE FEATURES, if overfitting occurs, then you can reduce the number of nodes.**</span>

## 3. Batch size

- Smaller batch size is slow, but generalizes the model better and can lead to better performance on the new data (test set).

- Larger batch size is faster, but can lead to overfitting and poor generalization on the new data (test set).

- So, to improve performance for larger batch size, scientiests came up with techniques like **Warming up the learning rate** / **Learning rate scheduling**.

    - In this technique, **for initial few epochs, we start with a smaller learning rate and gradually increase it to the desired learning rate. This can help the model to converge better and improve its performance.**

- ### So start with large batch size (as it is fast) if it doesn't perform well, then try smaller batch size as it will generalize better for sure.
