# What are activation functions?



- Activation function is acting like a gate between the input and output of a neuron. It decides whether a neuron should be activated or not by calculating a weighted sum and further adding bias with it.

# Important of activation functions



- If we do not use activation function in a neural network, then it will behave like a linear regression model. It will not be able to learn complex data such as images, videos, and audio. 

- Using activation functions, we can introduce non-linearity in the model. It helps the model to learn complex data and make better predictions.





- You can see in the above function if we take $g(z) = z$, as the activation function, then the output will be a linear function of the input.

- But instead if we take activation function like sigmoid, then the output will be a non-linear function of the input.

- Hence, it is said to use non-linear activation functions in the hidden layers of a neural network to capture complex patterns in the data.

# Quality of Ideal Activation Function

1. **Non-linearity**: The activation function should introduce non-linearity to the model, allowing it to learn complex (non-linear) patterns in the data.
    - The **Universal Approximation Theorem** states that a feedforward neural network with at least one hidden layer and non-linear activation functions can approximate any continuous function to a desired degree of accuracy, given sufficient neurons in the hidden layer. This theorem highlights the importance of non-linear activation functions in enabling neural networks to model complex relationships in data.

2. **Differentiability**: The activation function should be differentiable, allowing the use of gradient-based optimization methods (like backpropagation) to update the model's weights during training.
    - This is preferable but not strictly necessary. Some activation functions, like ReLU, are not differentiable at certain points (e.g., at zero), but they are still widely used due to their practical benefits.

3. **Computational Efficiency**: The activation function should be computationally efficient to evaluate, as it will be applied to many neurons in the network during both training and inference so computational efficiency is important for training speed and scalability.

4. **Zero-Centered**: The activation function should ideally be zero-centered, meaning that its output should have a mean of zero. i.e. The output should be normalized
    - NN converges faster if the input provided are normalized. 

5. **Non-saturating**: The activation function should avoid saturation (squeezing values in arange), where the output becomes insensitive to changes in the input. Saturation can lead to **vanishing gradients**, making it difficult for the model to learn effectively.
    - For example, the sigmoid and tanh functions can saturate for large positive or negative inputs, leading to very small gradients and slow learning. ReLU and its variants are often preferred because they do not saturate for positive inputs.

# Types of Activation Functions

## 1. Sigmoid Activation Function



- You can see that the derivative when recives a value outside range of (-3, 3) it almost becomes zero. This can lead to vanishing gradients problem

### Advantages:

1. Outputs values in the range (0, 1), making it suitable for modeling probabilities. And can be used in the output layer for binary classification problems.

2. Non-linear, allowing the model to learn complex patterns in the data.

3. Smooth and differentiable, enabling the use of gradient while backpropagation for training the model.

### Disadvantages:

1. It is a saturating activation function, converts ($-\infty$ , $\infty$) input to (0 , 1), which can lead to vanishing gradients for large positive or negative inputs. 
    - In vanishing gradients, the gradients become very small that the weights stop updating, and the model stops learning.
    - So, thats why sigmoid is not used in hidden layers of deep neural networks. It is used in the output layer for binary classification problems.

2. It is not zero-centered, mean the output that comes out of a layer of neuron containing sigmoid activation function will always be positive. Hence never be zero-centered.
    - This leads to slow training and convergence of the model, as the gradients will always have the same sign, causing oscillations in the weight updates.





- So you can see the $a_{21}$ and $a_{22}$ values are output ferom the previous layer and are always positive as a result of the sigmoid function.

- So, depending on the red boxed gradients in the above pic, bothe the final gradient can either be positive or negative, but the gradients will always have the same sign 



- So for a particular layer, the gradients will always have the same sign, <text style="color:orange"> hence bringing restrictions to the weight updates and causing oscillations in the optimization process, leading to slow convergence of the model.</text>

- So while minimizing the loss function we have constraints that gradients will always have the same sign, hence we move in a zig-zag manner towards the minima, which is not optimal and leads to slow convergence of the model.

3. Computationally expensive, as it involves calculating the exponential function, which can be slow for large inputs.

## 2. Tanh Activation Function

- Outputs values in the range (-1, 1), making it zero-centered and suitable for hidden layers of neural networks.



- It only solves the problem of non-zero centered output of sigmoid function

### Advantages:

1. Non-linear, allowing the model to learn complex patterns in the data.
2. Smooth and differentiable, enabling the use of gradient while backpropagation for training the model.
3. Zero-centered as it outputs values in the range (-1, 1), which helps in the training of neural networks by providing a mean of zero for the output.

### Disadvantages:

1. It is a saturating activation function, converts ($-\infty$ , $\infty$) input to (-1 , 1), which can lead to vanishing gradients for large positive or negative inputs. 

2. Computationally expensive, as it involves calculating the exponential function, which can be slow for large inputs.
