# Problem with Perceptron Trick

- The weights and bias that we get from the perceptron trick are not necessarily the optimal weights and bias to classify b/w the two classes.



- Perceptron trick se hume above image mei se koi si vhi line mil skti hai jo dono classes ko separate krti hai, so hum bata nhi skte ki hume best line mil rhi hai ya nhi, so it is not guaranteed that we will get the optimal weights and bias.

- There can be problem in convergence as well, because we are choosing the points randomly to update the weights and bias, so think of a scenario where we only chose the corectly classifiec points, so it will never update the weights and bias and it will never converge (though this is a very rare scenario but it can happen).

- So, Perceptron trick is a good jugad to find the weights and bias but it is not guaranteed to give us the optimal weights and bias and it can also have problem in convergence.

#### <span style="color:orange;"> So in ML we use loss functions, not jugads, to find the optimal weights and bias. Because of loss fucntion we can quantify how good is one model from the other </span>



# Perceptron Loss Function



- Without regularization, the perceptron loss function is given by:

$$L(w, b) = \frac{1}{N} \sum_{i=1}^{N} max(0, -y_i (w \cdot x_i + b))$$

- It is also known as the hinge loss function, and it is used to train the perceptron model.



- So we need to find the weights and bias that minimize the above loss function, to do this we use gradient descent

# Geometric Intuition of Perceptron Loss Function





- Note: here we gave -1 to the unplaced points and +1 to the placed points

- So, the loss function is calcualting the distance (or kind of distance {$w_1x_1 + w_2x_2 + b$}) of the **misclassified points** from the decision boundary, and we are trying to minimize this distance by updating the weights and bias.

- So, if the point is correctly classified then it have no contribution to the loss function, but if the point is misclassified then it will have a positive contribution to the loss function, and we are trying to minimize it

- Now we have to apply gradient descent to update the weights and bias, so we need to calculate the gradient of the loss function with respect to the weights and bias.

# More Loss Functions



- As we discussed earlier, perceptron is a mathematical model, where:
    - Input comes from the data
    - We sum the input with some weights and bias
    - We apply an activation function to the sum to get the output
    - While training the model we need to find the optimal weights and bias, to do this we use loss.

- The best part is perceptron is very flexible from its design perspective.

- We can change the activation function according to our need, and we can also change the loss function according to our need.

    - If someone wants probabilistic output then we can use sigmoid activation function and binary cross entropy loss function

    - If someone wants multi class classification then we can use softmax activation function and categorical cross entropy loss function
    
    - If someone wants regression then we can use linear activation function **(means jo $z$ ka value hai whi return karega)** and mean squared error loss function, and so on.

- #### People say perceptron is similar to logistic regression, but it is only true when we use sigmoid activation function and binary cross entropy loss function, otherwise it is not true, because perceptron is a very flexible model, we can change the activation function and loss function according to our need.




- So, its just about changing the activation function and loss function, and we can get different models, but the underlying architecture is still the same, which is perceptron.
