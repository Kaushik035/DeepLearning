# Dropout Layers

- Dropout layers are a regularization technique used in neural networks to prevent overfitting.

- It is also know as Regularization by randomization



- Neural networks are prone to overfitting, because of their complexity

- For eg. ANN have multiple layers of neurons, and each neuron has many parameters (weights and biases) that can be adjusted during training. This can lead to the model learning the noise in the training data rather than the underlying patterns.

## Possible Solutions to Overfitting

1. **Add more training data**: Increasing the size of the training dataset can help the model generalize better and reduce overfitting.

2. **Reduce model complexity**: Simplifying the architecture of the neural network by reducing the number of layers or neurons can help prevent overfitting.

3. **Early stopping**: Monitoring the model's performance on a validation set during training and stopping when the performance starts to degrade can help prevent overfitting.

4. **Regularization techniques**: Techniques like L1 and L2 regularization can be applied to the loss function to penalize large weights, which can help prevent overfitting.

5. **Dropout layers**: Dropout is a regularization technique that randomly sets a fraction of the input units to zero during training, which helps prevent the model from becoming too reliant on specific neurons and encourages it to learn more robust features.

# The concept of Dropout Layers


- Suppose we are solving a binary classification problem with 5 input features.

- In the above picture. in the LHS we have a fully connected neural network with 5 input features, 2 hidden layers, and 1 output layer.

- In the RHS we have a neural network with dropout layers. Where at each epoch we randomly switch off some neurons in the input layer and hidden layers. 

- **Note**: At each epoch, the neurons that are switched off are different, essentially creating a different architecture for the neural network at each epoch.

# Why this works?

- Overfitting happens because of multi-layered neurons that have many parameters. The model can learn the noise in the training data rather than the underlying patterns.

- So to reduce overfitting, we can do two things:
    1. Reduce the number of neurons
    2. Do something to make the node not focus on a specific pattern. Rather focus on the overall pattern.

- And both the above points are achieved by dropout layers.



- For dropout_ratio = 0.5, we randomly switch off 50% of the neurons in the input layer and hidden layers at each epoch.

- So the 1.st point is achieved because we are switching off a percentage of the neurons in the input layer and hidden layers at each epoch, effectively reducing the number of neurons that are active at each epoch.



- Suppose as in the above pic, purple neuron have the highest weight with the yellow neuron. So the model will focus on the purple neuron and yellow neuron and learn the pattern between them. 

- But in epoch 2, the yellow neuron is switched off. So even if based on data the purple neuron has the highest weight with the yellow neuron, the model will not be able to learn that pattern because the yellow neuron is switched off.

- Hence the input from the previous layers will be much balanced and the model will not focus on a specific pattern. Rather it will focus on the overall pattern. Thus the 2.nd point is achieved.

- These are the main 2 reasons why dropout layers work and help reduce overfitting in neural networks.

# Random Forest Analogy

- As in random forest with column sampling, we randomly select a subset of features to train each decision tree. And create ensemble of decision trees to make predictions. This helps reduce overfitting and improve generalization.



- Similarly for the neural network, with 1 input layer and 1 hidden layer with 4 nodes each, we can make $2^8$ different architectures of the neural network by randomly switching off some neurons in the input layer and hidden layer at each epoch.

- so, if we are running our NN for 10 epochs, it is equivalent to training 10 different architectures of the neural network. And at the end of training, we will have an ensemble of 10 different architectures of the neural network to make predictions. This helps reduce overfitting and improve generalization.

# How Prediction works?

- <text style="color:orange"> Dropout bss training k time lagate hai, prediction k time nahi lagate hai</text>

- While prediction, all the weights and biases are used



- But suppose, while training a node was not available for all the epochs, then how will we count for its final weight value during prediction?



- So, what ever the weight values are learned during training, we multiply them with the (1 - $p$) to get the final weight values that will be used during prediction.

- $p$ = dropout ratio, it can be read as the probability of a neuron being switched off during training for a given layer.

- So, 1 - $p$ = probability of a neuron being active during training for a given layer.

- Hence we multiply the weight values learned during training with (1 - $p$) to get the final weight values that will be used during prediction. Because it is the probability of a neuron being active during training for a given layer.

# Effect of p



- As we increase the dropout ratio, the model will be more regularized and less likely to overfit. But if we increase the dropout ratio too much, the model will not be able to learn the underlying patterns in the data and will underfit.

# Practical Tips and Tricks

1. If overfitting then increase the dropout ratio and if underfitting then decrease the dropout ratio.

2. Start implementing dropout from the last hidden layer and then while observing the training and validation loss, implement dropout to the previous hidden layers.

3. For ANN, $p$ -> [0.1,0.5], For CNN, $p$ -> [0.4,0.5], For RNN, $p$ -> [0.2,0.3]



# Drawback of Dropout Layers

1. There comes a delay in convergence to reach the optimal weight in the NN

2. The value of the Cost function changes, as the dropout layer is randomly switching off some neurons at each epoch, the cost function will be different at each epoch. This makes gradient calculation and monitoring the training process more challenging, as the loss may fluctuate significantly from epoch to epoch.
